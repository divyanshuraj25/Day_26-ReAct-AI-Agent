```python
from google import genai
from getpass import getpass

GEMINI_API_KEY = getpass("Enter your Gemini API Key: ")

client = genai.Client(api_key=GEMINI_API_KEY)

MODEL = "gemini-3.6-flash"

print("Gemini API connected successfully!")
```

```python
from datetime import datetime
import ast
import operator as op

# ---------------- CALCULATOR ----------------

_ALLOWED_OPS = {
    ast.Add: op.add,
    ast.Sub: op.sub,
    ast.Mult: op.mul,
    ast.Div: op.truediv,
    ast.Pow: op.pow,
    ast.Mod: op.mod,
    ast.USub: op.neg,
    ast.UAdd: op.pos,
}

def calculator(expression):
    """Safely evaluate basic arithmetic."""

    def evaluate(node):
        if isinstance(node, ast.Constant) and isinstance(node.value, (int, float)):
            return node.value

        if isinstance(node, ast.BinOp):
            left = evaluate(node.left)
            right = evaluate(node.right)
            return _ALLOWED_OPS[type(node.op)](left, right)

        if isinstance(node, ast.UnaryOp):
            return _ALLOWED_OPS[type(node.op)](evaluate(node.operand))

        raise ValueError("Unsupported expression")

    return evaluate(ast.parse(expression, mode="eval").body)


# ---------------- KNOWLEDGE BASE ----------------

knowledge_base = """
Day 11 Knowledge Base

Python is a high-level programming language known for simple syntax and
large libraries.

Machine Learning is a subset of AI where systems learn patterns from data.

Natural Language Processing (NLP) deals with understanding and processing
human language.

A neural network consists of interconnected computational units called neurons.

RAG stands for Retrieval-Augmented Generation. It retrieves relevant
information before generating an answer.

Vector databases store numerical representations called embeddings and
allow similarity search.

AI agents can use tools, observe their results, and decide what action
to take next.
"""


def search_docs(query):
    """Simple keyword-based retrieval from Day 11 knowledge base."""

    query_words = set(query.lower().split())

    sentences = [
        s.strip()
        for s in knowledge_base.split("\n")
        if s.strip()
    ]

    scored = []

    for sentence in sentences:
        words = set(sentence.lower().split())
        score = len(query_words.intersection(words))

        if score > 0:
            scored.append((score, sentence))

    scored.sort(reverse=True)

    if not scored:
        return "No relevant document found."

    return "\n".join(sentence for score, sentence in scored[:3])


# ---------------- TODAY ----------------

def get_today():
    return datetime.now().strftime("%Y-%m-%d")


print("Tools created:")
print("- calculator()")
print("- search_docs()")
print("- get_today()")
```

```python
tool_registry = {
    "calculator": calculator,
    "search_docs": search_docs,
    "get_today": get_today
}


def dispatch(action, parameters):
    """
    Execute the tool selected by the LLM.
    """

    if action not in tool_registry:
        return f"ERROR: Unknown tool '{action}'"

    try:
        result = tool_registry[action](**parameters)
        return str(result)
    except Exception as e:
        return f"ERROR while executing {action}: {e}"


print(tool_registry)
```

```python
import json
import re


SYSTEM_PROMPT = """
You are a manual ReAct AI agent.

You have exactly three tools:

1. calculator(expression)
2. search_docs(query)
3. get_today()

You must solve problems using this format:

Thought: explain what needs to be done.
Action: tool_name
Parameters: JSON parameters

After receiving an Observation, continue reasoning.

When the task is completely solved:

Final Answer: answer the user clearly.

IMPORTANT:
- Never invent an Observation.
- Only use information returned by tools.
- Use multiple tools when necessary.
- Do not use tools unnecessarily.
- Stop when the answer is complete.

Valid actions:
calculator
search_docs
get_today
"""


def extract_action(text):
    """
    Extract Action and Parameters from Gemini response.
    """

    action_match = re.search(
        r"Action:\s*([a-zA-Z_]+)",
        text,
        re.IGNORECASE
    )

    params_match = re.search(
        r"Parameters:\s*(\{.*?\})",
        text,
        re.DOTALL
    )

    if not action_match:
        return None, {}

    action = action_match.group(1).strip()

    if params_match:
        try:
            parameters = json.loads(params_match.group(1))
        except:
            parameters = {}
    else:
        parameters = {}

    return action, parameters


def run_agent(question, max_steps=8):

    conversation = f"""
{SYSTEM_PROMPT}

User Question:
{question}
"""

    trace = []

    for step in range(1, max_steps + 1):

        response = client.models.generate_content(
            model=MODEL,
            contents=conversation
        )

        output = response.text.strip()

        print("\n" + "=" * 70)
        print(f"STEP {step}")
        print("=" * 70)
        print(output)

        trace.append({
            "step": step,
            "llm_output": output
        })

        # Final answer
        if "Final Answer:" in output:
            return output, trace

        action, parameters = extract_action(output)

        if action is None:
            observation = "ERROR: No valid Action found."

        else:
            print(f"\nExecuting tool: {action}")
            print(f"Parameters: {parameters}")

            observation = dispatch(action, parameters)

            print(f"Observation: {observation}")

        trace[-1]["action"] = action
        trace[-1]["parameters"] = parameters
        trace[-1]["observation"] = observation

        conversation += f"""

Assistant:
{output}

Observation:
{observation}

Continue the ReAct process.
"""

    return "Agent stopped because maximum steps were reached.", trace
```

```python
answer, trace = run_agent(
    "What is 125 * 8 + 50?"
)

print("\nFINAL RESULT")
print(answer)
```

```python
# Replace your current run_agent() with this version

def run_agent(question, max_steps=8):

    conversation = f"""
{SYSTEM_PROMPT}

User Question:
{question}
"""

    trace = []

    for step in range(1, max_steps + 1):

        try:
            response = client.models.generate_content(
                model=MODEL,
                contents=conversation
            )

        except Exception as e:
            error_text = str(e)

            print("\n" + "=" * 70)
            print(f"API ERROR - STEP {step}")
            print("=" * 70)
            print(error_text)

            trace.append({
                "step": step,
                "error": error_text
            })

            return {
                "status": "API_ERROR",
                "answer": None,
                "trace": trace
            }

        output = response.text.strip()

        print("\n" + "=" * 70)
        print(f"STEP {step}")
        print("=" * 70)
        print(output)

        # Final answer
        if "Final Answer:" in output:
            trace.append({
                "step": step,
                "llm_output": output,
                "status": "completed"
            })

            return {
                "status": "completed",
                "answer": output,
                "trace": trace
            }

        action, parameters = extract_action(output)

        if action is None:
            observation = "ERROR: No valid Action found."

        else:
            print(f"\nExecuting tool: {action}")
            print(f"Parameters: {parameters}")

            observation = dispatch(action, parameters)

            print(f"Observation: {observation}")

        trace.append({
            "step": step,
            "llm_output": output,
            "action": action,
            "parameters": parameters,
            "observation": observation
        })

        conversation += f"""

Assistant:
{output}

Observation:
{observation}

Continue the ReAct process.
"""

    return {
        "status": "MAX_STEPS_REACHED",
        "answer": None,
        "trace": trace
    }


print("Updated run_agent() successfully.")
```
